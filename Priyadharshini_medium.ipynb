```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

sparkSession = SparkSession.builder.appName("EmployeeSAssgn").master("local[*]").getOrCreate()
print(sparkSession)
sc = sparkSession.sparkContext
```

```python
rdd = sc.parallelize([
(101, "Arun", "IT", 45000),
(102, "Divya", "HR", 38000),
(103, "Kavin", "IT", 62000),
(104, "Meena", "Finance", 55000),
(105, "Ravi", "IT", 29000),
(106, "Priya", "HR", 48000),
(107, "Sanjay", "Finance", 72000),
(108, "Nisha", "IT", 51000),
(109, "Vijay", "HR", 32000),
(110, "Anu", "Finance", 42000)
])

result = rdd.filter(lambda x: x[2] == "IT" and x[3] > 40000)
result = result.map(lambda x: f"Employee: {x[1]} | Department: {x[2]} | Original Salary: {x[3]} | Revised Salary: {int(x[3] * 1.10)}")
print("Employee Salary Processing:")
for employee in result.collect():
    print(employee)
```

```python
student_data = [
    (1, "Arun", 85, 78, 92),
    (2, "Divya", 45, 55, 60),
    (3, "Kavin", 35, 40, 38),
    (4, "Meena", 90, 88, 95),
    (5, "Ravi", 65, 72, 58),
    (6, "Priya", 30, 85, 80),
    (7, "Sanjay", 55, 60, 65),
    (8, "Nisha", 95, 92, 89),
    (9, "Vijay", 40, 35, 70),
    (10, "Anu", 75, 80, 85)
]

columns = ["student_id", "name", "python", "sql", "pyspark"]
student_df = sparkSession.createDataFrame(student_data, columns)
student_df = student_df.withColumn(
    "total_marks", col("python") + col("sql") + col("pyspark")
)
student_df = student_df.withColumn(
    "average_marks",(col("total_marks") / 3)
)
student_df = student_df.withColumn(
    "result",
    when(
        (col("python") >= 40) & (col("sql") >= 40) & (col("pyspark") >= 40),
        "PASS"
    ).otherwise("FAIL")
)
student_df = student_df.withColumn(
    "grade",
    when(col("average_marks") >= 90, "A")
    .when(col("average_marks") >= 75, "B")
    .when(col("average_marks") >= 60, "C")
    .otherwise("D")
)
print("All students")
student_df.select(
    "name", "total_marks", "average_marks", "result", "grade"
).show()
```

```python
student_df.filter((col("result") == "PASS") & (col("average_marks") >= 75)).select(
    "name", "total_marks", "average_marks", "result", "grade"
).show()
```
